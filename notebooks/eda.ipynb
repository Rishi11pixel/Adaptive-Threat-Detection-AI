```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/processed_sensor_data.csv")

normal = df[df["label"] == 0]
anomaly = df[df["label"] == 1]
```

```python
plt.figure()
plt.hist(normal["motion"], bins=50, alpha=0.7, label="Normal")
plt.hist(anomaly["motion"], bins=50, alpha=0.7, label="Anomaly")
plt.legend()
plt.title("Motion Sensor Distribution")
plt.show()
```
